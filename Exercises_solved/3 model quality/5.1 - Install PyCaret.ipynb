# Install PyCaret

Normally speaking we wouldn't be creating a new notebook for this as it would boil down to "pip install pycaret". The problem is that pycaret doesn't always work with the latest version of Python. That means it may be neccessary to create a new virtual environment after installing an older version of Python. Even when you are still using Python 3.11 as your main python (which would be fine, as it still receives security updates) it's advisable to install a new venv just for PyCaret to avoid potential conflicts with other packages.

When installing the older version of Python, make sure to **not** add it to the path, as that may overwrite your newer Python-version.

First step, open up a terminal and run

```Shell
python --version
```
Now check if this version is on [this](https://pycaret.gitbook.io/docs/get-started/installation) webpage.

If it isn't first install an older Python version and keep track of the installation location as we'll need that to create a venv with that older version.

Go back to your terminal and create a new venv in the working folder of this course.
```Shell
."C:\Program Files\Python311\python.exe" -m venv venv_pycaret
```

This will create a new venv in this folder, called "venv_pycaret". Make sure to select it when running the following cell.

Fun fact: When using Python 3.12 PyCaret will happily install over the course of several minutes. When you run it, however, you'll get:
